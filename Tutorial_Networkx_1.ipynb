 **ALGORITMO DE ÁRBOL DE EXPANSIÓN MÍNIMA**

El algoritmo de árbol de expansión mínima enlaza los nodos de una red, en forma directa o indirecta, con la mínima longitud de las ramas enlazantes.

Los pasos del procedimiento son los siguientes. Sea $N = \{1, 2, \dots, n\}$ el conjunto de nodos de la red, y se definen:

* $C_k =$ Conjunto de nodos que se han conectado en forma permanente en la iteración $k$
* $\bar{C}_k =$ Conjunto de nodos que todavía se deben conectar en forma permanente

**Paso 0.** El conjunto $C_0 = \emptyset$ y $\bar{C}_0 = N$.

**Paso 1.** Comenzar con *cualquier nodo* en el conjunto $\bar{C}_0$ no conectado (o "inconexo"), e igualar $C_1 = \{i\}$, con lo que $\bar{C}_1 = N - \{i\}$. Igualar $k = 2$.

**Paso general k.** Seleccionar un nodo $j^*$ en el conjunto no conectado $\bar{C}_{k-1}$ que produzca el arco más corto a un nodo, en el conjunto conectado $C_{k-1}$. Enlazar a $j^*$ en forma permanente con $C_{k-1}$ y sacarlo de $\bar{C}_{k-1}$, esto es

$$C_k = C_{k-1} + \{j^*\}, \bar{C}_k = \bar{C}_{k-1} - \{j^*\}$$

Si el conjunto $\bar{C}_k$ de nodos no conectados es vacío, detenerse. En cualquier otro caso, igualar $k = k + 1$ y repetir el paso.

```python
import matplotlib.pyplot as plt #Biblioteca para mostrar  la red
import networkx as nx # Biblioteca  para crear la red, manejar sus nodos, conexiones y pesos

# Crea la red no dirigida
G = nx.Graph()

#Lista de conexiones entre nodos , con su respectivo peso
conexiones = [(1, 2, 5), (1, 3, 1), (3, 2, 2), (2, 4, 7), (2, 5,1 ),(2, 6, 6), (6, 2, 7), (3, 4, 6), (4, 3, 7), (4, 6, 4),(4, 7, 6), (3, 5, 7), (5, 4, 3), (5, 6, 5), (5, 7, 9),(6, 7, 2),]
G.add_weighted_edges_from(conexiones)#Método rápido que añade todas conexiones entre nodos con sus respectivos pesos
E = nx.minimum_spanning_tree(G, weight="weight")#Funcion para obtener el arbol de expansion minima

print("Las  Conexiones del Árbol de Expansión Mínima son ")
print(E.edges(data=True))#obtiene los pesos del árbol

# Fijamos manualmenta la ubicación de cada nodo ,para evitar que se coloquen de forma desordenada
pos  = {
    1: (0, 1),  # Extremo izquierdo
    2: (1, 2),  # Arriba izquierda
    3: (1, 0),  # Abajo izquierda
    4: (3, 1),  # Centro
    5: (3, 0),  # Abajo derecha
    6: (3, 2),  # Arriba derecha
    7: (4, 1),  # Extremo derecho
}

# Dibuja el árbol E,  pos da las coordenadas de cada nodo para saber donde colocar cada nodo
#se escribe el numero de cada nodo y se asigna un tamaño a los nodos
nx.draw(E, pos, with_labels=True, node_color="cyan", node_size=700,edgecolors="black",font_size=15,font_weight="bold")

# Extrae  un diccionario que contiene los pesos del arbol de expansion minima
pesos = nx.get_edge_attributes(E, "weight")

# Escribe  los pesos sobre las conexiones
nx.draw_networkx_edge_labels(E, pos, edge_labels=pesos,font_size=15)

# Muestra el árbol de expansión mínima
plt.title(
    "Árbol de Expansión Mínima", fontsize=12, fontweight="bold", pad=20
)
plt.show()
```

<font color='Purple' face="Arial" size="6"><b>RUTA MÁS CORTA </b></font>

<font color='#374151' face="Segoe UI" size="3">Para obtener la Ruta más corta ,utilizaremos las siguientes funciones:


* **nx.shortest_path(G, source, target, weight):** Devuelve la secuencia de nodos que forman el camino más corto entre un nodo inicial (source) y uno final (target).
*  **nx.shortest_path_length(G, source, target, weight):** Calcula únicamente la distancia numérica total o el costo de la ruta más corta.

***Parámetros:***
* **G:** El grafo que creaste (puede ser dirigido o no dirigido).
* **source**: El nodo de inicio (origen).
* **target**: El nodo final (destino). Si lo omites, calcula la ruta hacia todos los nodos.
*  **weight**: El nombre del peso de las aristas (ej: 'weight', 'distancia'). Si lo omites, cada arista vale 1.
*  **method**: El algoritmo a usar ('dijkstra' por defecto). Se cambia a 'bellman-ford' si hay pesos negativos.

Las funciones necesitan acceder al objeto G y a los pesos de las aristas, por lo que deben ejecutarse después de ***G.add_weighted_edges_from(conexiones).***

</font>

**Algoritmo de la ruta más corta**

* **Objetivo de la $n$-ésima iteración:** encontrar el $n$-ésimo nodo más cercano al origen. (Este paso se repetirá para $n = 1, 2, \dots$ hasta que el $n$-ésimo nodo más cercano sea el nodo destino.)

* **Datos de la $n$-ésima iteración:** $n - 1$ nodos más cercanos al origen —que se encontró en las iteraciones previas—, incluida su ruta más corta y la distancia desde el origen. (Estos nodos y el origen se llaman *nodos resueltos*; el resto son nodos *no resueltos*.)

* **Candidatos para $n$-ésimo nodo más cercano:** cada nodo resuelto que tiene conexión directa por una ligadura con uno o más nodos no resueltos proporciona *un candidato*, esto es, el nodo no resuelto que tiene la ligadura más *corta*. (Los empates proporcionan candidatos adicionales.)

* **Cálculo del $n$-ésimo nodo más cercano:** para cada nodo resuelto y sus candidatos, se suma la distancia entre ellos y la distancia de la ruta más corta desde el origen a este nodo resuelto. El candidato con la distancia total más pequeña es el $n$-ésimo nodo más cercano —los empates proporcionan nodos resueltos adicionales—, y su ruta más corta es la que genera esta distancia.

<font color='#374151' face="Segoe UI" size="4">**Ahora aplicamos la función para obtener la ruta más corta.**</font>

```python
import matplotlib.pyplot as plt  # # Biblioteca para mostrar  la red
import networkx as nx  # Biblioteca  para crear la red, manejar sus nodos, conexiones y pesos

#Crea la red dirigida
G = nx.DiGraph()

# Definimos las conexiones  (Origen, Destino, Peso)
conexiones = [(1, 2, 5), (1, 3, 1), (3, 2, 2), (2, 4, 7), (2, 5,1 ),(2, 6, 6), (6, 2, 7), (3, 4, 6), (4, 3, 7), (4, 6, 4),(4, 7, 6), (3, 5, 7), (5, 4, 3), (5, 6, 5), (5, 7, 9),(6, 7, 2),]

# Método que añade automáticamente todos los nodos y conexiones con sus respectivos pesos
G.add_weighted_edges_from(conexiones)
#--------------------------------------------------------------------------------------------------
#---------------------CALCULO DE LA RUTA MAS CORTA-------------------------------------------------
origen=1#Definimos el nodo de origen
destino=7#Definimos el nodo de destino
#Utilizamos la funcion para obtener la secuencia de nodos que conforman la ruta mas corta
ruta_corta=nx.shortest_path(G, source=origen, target=destino, weight="weight")
#Calculamos la distancia de la ruta más corta
distancia_min=nx.shortest_path_length(G, source=origen, target=destino, weight="weight")
#Imprimimos los resultados
print('LA RUTA MÁS CORTA ES',ruta_corta)
print('LA DISTANCIA MÍNIMA ES',distancia_min)



#Creamos un diccionario que contien las posiciones de cada nodo, las cuales se eligieron de forma manual.
pos = {
    1: (0, 2),  # Extremo izquierdo
    2: (2, 4),  # Arriba izquierda
    3: (2, 0),  # Abajo izquierda
    4: (3.5, 2),  # Centro de la red
    5: (7.3, 0),  # Abajo derecha
    6: (7, 4),  # Arriba derecha
    7: (8, 2),  # Extremo derecho
}

curvas_arriba = [(2, 6), (3, 4)]  # Lista de las flechas que se enviarán hacia un arco
curvas_abajo = [(6, 2), (4, 3)]   # Lista de las flechas que se enviarán hacia el arco contrario

# Lista  para filtrar todas las conexiones que se mantendrán  rectas
rectas = [e for e in G.edges() if e not in curvas_arriba + curvas_abajo]

#Definimos el tamaño de la red
plt.figure(figsize=(12, 7))

# Dibujamos los nodos , borde negro y tamaño de 1200
nx.draw_networkx_nodes(G, pos, node_color="plum", node_size=1200, edgecolors="black")

# Escribimos los números de los nodos en el centro de cada círculo
nx.draw_networkx_labels(G, pos, font_size=12, font_weight="bold")

# Dibujamos las flechas RECTAS
nx.draw_networkx_edges(G, pos, edgelist=rectas, arrows=True, arrowsize=20, node_size=1200)

# Dibuja las flechas CURVAS superiores con un radio de curvatura de -0.2
nx.draw_networkx_edges(
    G, pos, edgelist=curvas_arriba, arrows=True, arrowsize=20, node_size=1200,
    connectionstyle="arc3,rad=-0.2"
)

# Dibuja las flechas CURVAS inferiores aplicando el mismo radio
nx.draw_networkx_edges(
    G, pos, edgelist=curvas_abajo, arrows=True, arrowsize=20, node_size=1200,
    connectionstyle="arc3,rad=-0.2"
)


# Extraemos un diccionario con los valores de los pesos
pesos = nx.get_edge_attributes(G, "weight")

#  Dibujamos los pesos de las conexiones  RECTAS:
# - label_pos=0.5 las coloca exactamente en la mitad de la arista.
# - rotate=False mantiene los números legibles horizontalmente.
# - bbox crea un fondo blanco semitransparente con esquinas redondeadas para no pisar las líneas.
pesos_rectas = {e: pesos[e] for e in rectas}
nx.draw_networkx_edge_labels(
    G, pos, edge_labels=pesos_rectas, label_pos=0.5, font_size=15, rotate=False,
    bbox=dict(boxstyle="round,pad=0.2", fc="white", ec="none", alpha=0.8)
)

# Dibujamos los pesos de las aristas CURVAS SUPERIORES:
# Al pasarle el mismo 'connectionstyle', NetworkX desplaza la etiqueta junto con la trayectoria del arco.
pesos_c_arriba = {e: pesos[e] for e in curvas_arriba}
nx.draw_networkx_edge_labels(
    G, pos, edge_labels=pesos_c_arriba, label_pos=0.5, font_size=15, rotate=False,
    connectionstyle="arc3,rad=-0.2",
    bbox=dict(boxstyle="round,pad=0.2", fc="white", ec="none", alpha=0.8)
)

# Dibujamos los pesos de las aristas CURVAS INFERIORES
pesos_c_abajo = {e: pesos[e] for e in curvas_abajo}
nx.draw_networkx_edge_labels(
    G, pos, edge_labels=pesos_c_abajo, label_pos=0.5, font_size=15, rotate=False,
    connectionstyle="arc3,rad=-0.2",
    bbox=dict(boxstyle="round,pad=0.2", fc="white", ec="none", alpha=0.8)
)

# Mostramos la red
plt.axis("off")       # Oculta los ejes X e Y del plano cartesiano
plt.tight_layout()   # Ajusta los márgenes para que no se corte el gráfico
plt.show()           # Renderiza la ventana con el grafo resultante
```

<font color='#374151' face="Segoe UI" size="3">
Utilizaremos la función  nx.maximum_flow(), que se utiliza para calcular el flujo máximo de una red (grafo dirigido con capacidades) desde un nodo de origen  hasta un nodo de destino .


</font>

Síntaxis básica:
**nx.maximum_flow(G, s, t, capacity='capacity', flow_func=None)**<font color='#374151' face="Segoe UI" size="3">


• G (NetworkX DiGraph):Representa la red.

• s (nodo):El nodo de origen  
• t (nodo):El nodo de destino  
• capacity (string, opcional):Nombre del atributo de las aristas que contiene el valor de la capacidad. Por defecto busca la palabra 'capacity'. Si una arista no tiene este atributo, se asume que su capacidad es infinita.

• flow_func (función, opcional):Permite elegir el algoritmo específico que realizará el cálculo. Si se deja en None (por defecto), NetworkX elegirá automáticamente el más eficiente según la estructura del grafo (usualmente edmonds_karp o preflow_push).


La función devuelve una tupla con dos elementos:
1. flow_value (entero o flotante): El valor numérico del flujo máximo que se puede enviar desde s hasta t.
2. flow_dict (diccionario de diccionarios): Un diccionario que detalla cuánto flujo pasa exactamente por cada arista. Por ejemplo, flow_dict[u][v] te dará el flujo enviado del nodo u al nodo v.

</font>

**Algoritmo**

1. Identificar una trayectoria de aumento (si no existe una,los flujos netos asignados constituyen un patrón de flujo óptimo)

2. Identificar la capacidad residual $c^*$ de esta trayectoria ,se aumenta en $c^*$ el flujo de esta trayectoria .

3. Se disminuye en  $c^*$ la capacidad residual de cada arco en esta trayectoria de aumento.Se aumenta en $c^*$ la capacidad residual de cada arco en la direccion opuesta en esta trayectoria.
Regresar al paso 1

```python
import matplotlib.pyplot as plt  # Biblioteca para mostrar  la red
import networkx as nx   # Biblioteca  para crear la red, manejar sus nodos, conexiones y pesos

# Crea la red dirigida
G = nx.DiGraph()

# Definimos las conexiones (Origen, Destino, Capacidad)
conexiones = [
    (1, 3, 14), (3, 1, 10),  # <--- Agrega (3, 1, 10)
    (1, 5, 4),  (5, 1, 0),
    (1, 2, 8),  (2, 1, 0),   # <--- Agrega (2, 1, 0)
    (2, 3, 5),  (3, 2, 0),
    (2, 5, 6),  (5, 2, 0),
    (2, 4, 7),  (4, 2, 6),
    (3, 5, 10), (5, 3, 0),
    (3, 4, 9),  (4, 3, 7),
    (4, 5, 5),  (5, 4, 0)
]

# Método que añade automáticamente todos los nodos y conexiones con su capacidad (capacidad = weight)
for u, v, w in conexiones:
    G.add_edge(u, v, capacity=w, weight=w)

# ==============================================================================
# CÁLCULO DEL FLUJO MÁXIMO
# ==============================================================================
origen = 1
destino = 5

# Calcula el valor del flujo máximo y el diccionario con la distribución del flujo por cada arista
valor_flujo_maximo, flujo_diccionario = nx.maximum_flow(G, origen, destino, capacity="capacity")

print(f"El Flujo Máximo del nodo {origen} al nodo {destino} es: {valor_flujo_maximo}")
print("Distribución del flujo por arista:")
for u in flujo_diccionario:
    for v in flujo_diccionario[u]:
        if G[u][v]["capacity"] > 0:  # Mostrar solo conexiones relevantes
            print(f"  Arista ({u} -> {v}): {flujo_diccionario[u][v]} / {G[u][v]['capacity']}")
# ==============================================================================

# #Creamos un diccionario que contien las posiciones de cada nodo, las cuales se eligieron de forma manual.
pos = {
    1: (0, 2),    # Extremo izquierdo
    2: (2, 0),    # Abajo izquierda
    3: (3, 4),    # Arriba centro
    4: (5, 0),    # Abajo derecha
    5: (7, 2),    # Extremo derecho
}

# Definimos el tamaño de la red
plt.figure(figsize=(10, 7))

# Dibujamos los nodos con fondo blanco, borde negro
nx.draw_networkx_nodes(G, pos, node_color="white", node_size=1500, edgecolors="black")

# Escribimos los números de los nodos en el centro ,en negritas
nx.draw_networkx_labels(G, pos, font_size=14, font_weight="bold")

# Dibujamos las conexiones de la red
nx.draw_networkx_edges(G, pos, arrows=False, edge_color="black", node_size=1500)

# Extraemos un diccionario con los valores de las capacidades para mostrarlos en el gráfico
pesos = nx.get_edge_attributes(G, "capacity")

# Dibujamos las etiquetas de las capacidades sobre las líneas
nx.draw_networkx_edge_labels(
    G, pos, edge_labels=pesos, label_pos=0.25, font_size=15, rotate=False,
    bbox=dict(boxstyle="round,pad=0.1", fc="white", ec="none", alpha=0.9)
)

# Mostramos la red
plt.axis("off")       # Oculta los ejes X e Y del plano cartesiano
plt.tight_layout()   # Ajusta los márgenes para que no se corte el gráfico
plt.show()           # Renderiza la ventana con el grafo resultante.
```
