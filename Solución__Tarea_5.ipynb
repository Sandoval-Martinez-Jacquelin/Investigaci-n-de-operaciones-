<h1><b><font color="red" >Solución del Problema 1 .Tarea 5 por Simulación de Monte Carlo</b></h1></font>


<p style="font-family: 'Arial', 'Helvetica', sans-serif; font-size: 16px; line-height: 1.6; color: #333333;">
  Se resolverá este problema utilizando simulación de Monte Carlo, entonces se realizará el código para hacer varias simulaciones, y se contará el número de veces en que la rata encontró la comida antes de recibir una descarga. Luego, se realizará la división:
</p>

<ul style="font-family: 'Arial', 'Helvetica', sans-serif; font-size: 15px; line-height: 1.6; color: #333333; margin-left: 20px;">
  <li><b>$n$:</b> Número de simulaciones donde la rata encontró la comida antes de recibir una descarga.</li>
  <li><b>$N$:</b> Número total de simulaciones.</li>
</ul>

$$ \frac{n}{N} \approx P(\text{La rata encuentre la comida antes de recibir una descarga}) $$


<p style="font-family: 'Arial', 'Helvetica', sans-serif; font-size: 16px; line-height: 1.6; color: #333333;">
Algunas consideraciones a tener en cuenta al realizar el código de la simulación son las siguientes:

  • Si desde el compartimiento actual $i$ existen $K$ salidas posibles , la rata elige cada una con probabilidad $1/K$.La probabilidad de pasar del compartimiento $i$ a uno que no tiene conexion directa con este ,es $0$.
</p>

<p style="font-family: 'Arial', 'Helvetica', sans-serif; font-size: 16px; line-height: 1.6; color: #333333;">
  • Si la rata llega al compartimiento $7$ se queda comiendo , si llega al $8$ recibe descargas electricas , entonces al llegar a cualquiera de los dos, el estado no puede cambiar y asi se tiene :
</p>

$$ P( X_{n+1} = 8 \mid X_n = 8 ) = P_{88} = 1 $$
$$ P( X_{n+1} = 7 \mid X_n = 7 ) = P_{77} = 1 $$

Ahora procederemos a escribir el código ,en cada línea se explicará lo que se esta realizando.Para obtener una buena aproximación ,realizaremos N=10000000 simulaciones.

```python
#Importamos la funcion 'choice' de la libreria 'random',la cual nos permitirá seleccionar un compartimiento de los k posibles ,con probabilidad 1/k
from random import choice

#Luego,creamos la función 'compartimientos_posibles', la cual nos dará como resultado una lista de los compartimientos
# a los que se puede mover la rata , dependiendo del compartimiento actual(argumento de la función)
def compartimientos_posibles(compartimiento_actual):
  #Obervanmos la imagen del laberinto y creamos la lista de los posibles compartimientos ,dependiendo del compartimiento actual donde esta la rata
  if compartimiento_actual==0:
    return [1,2]#Si la rata está en el compartimiento 0, solo se puede mover al 1 o al 2 en el siguiente paso
  elif compartimiento_actual==1:
    return [0,3,7]#Si la rata está en el compartimiento 1, solo se puede mover al 0,3 o 7 en el siguiente paso
  elif compartimiento_actual==2:
    return [0,3,8]#Si la rata está en el compartimiento 2, solo se puede mover al 0,3 u 8 en el siguiente paso
  elif compartimiento_actual==3:
    return [1,2,4,5]#Si la rata está en el compartimiento 3, solo se puede mover al 1,2,4,5 en el siguiente paso
  elif compartimiento_actual==4:
    return [3,6,7]#Si la rata está en el compartimiento 4, solo se puede mover al 3,6 o 7 en el siguiente paso
  elif compartimiento_actual==5:
    return [3,8,6]#Si la rata está en el compartimiento 5, solo se puede mover al  3,8 o 6 en el siguiente paso
  elif compartimiento_actual==6:
    return [4,5]#Si la rata está en el compartimiento 6, solo se puede mover al 4 o 5 en el siguiente paso
  elif compartimiento_actual==7:
    return [7]#Si la rata está en el compartimiento 7 ya no se puede mover a otro compartimiento,se queda comiendo
  else:
    return [8]#Si la rata está en el compartimiento 8 ya no se puede mover a otro compartimiento ,recibe una descarga electrica



#--------------------------------------------------------------------------------------------------------------------------------
#Ahora definimos una funcion , para UNA sola simulación , con un estado inicial fijo (puede ser:0,1,2,3,4,5,6)
def laberinto(estado_inicial):
  #Definimos el estado actual como el estado inicial donde comienza la rata
  estado_actual=estado_inicial
  #Este ciclo, simula los compartimientos que la rata va recorriendo y se detine cuando llega a los estados absorbentes que son el compartimiento 7 u 8
  while estado_actual!=7 and estado_actual!=8:
    #Luego hacemos que el estado pasado sea el estado actual ,definido anteriormente
    estado_pasado=estado_actual
    #De los k compartimientos posibles que existen( dependiendo del 'estado pasado') , la función 'choice' seleccionará uno ,a donde se moverá la rata.
    estado_actual=choice(compartimientos_posibles(estado_pasado))
#Una vez que la rata llega al compartimiento 7 u 8 , el ciclo se detine .
#La función nos devolverá un 1  si la rata en esta simulaçión se queda en el compartimiento 7(un estado absorbente), y un 0 si se queda en el 8
  if estado_actual==7:
    return 1
  else:
    return 0

#Definimos una lista de los estados transitorios
estados_transitorios=[0,1,2,3,4,5,6]
N=10000000 #Es el número de simulaciones a realizar, para obtener una mejor precisión de las probabilidades
p=0#contador para ir sumando las probabilidades obtenidas , dependiendo de cada estado transitorio de inicio
#Realizamos este ciclo para calcular la probabilidad de que la rata termine en el compartimiento 7 ,si la rata comenzó en cada uno de los estados tarnsitorios.
for m in estados_transitorios:
  n=0 #Contará el Número de simulaciones donde la rata encontró la comida antes de recibir una descarga,comenzando desde el estado transitorio 'm'
  for i in range(N):#Realiza N simulaciones
    n+=laberinto(m) #Se van sumando las veces en que la rata encuentra la comida antes de recibir una descarga,comenzando desde el estado transitorio 'm'
  #Por ultimo ,imprimimos la probabilidad calculada de que la rata encuentre la comida antes de recibir una descarga , comenzando en el estado transitorio 'm'
  print("La probabilidad de empezar en ",m," y terminar absorbido en 7 es",n/N)
  p+=n/N#se van sumando las probabilidades obtenidas , dependiendo de cada estado transitorio de inicio


#Imprimimos  la probabilidad en caso de que la rata se coloque de manera equiprobbale en cualquier compartimiento
print("Si la rata se coloca de manera equiprobable al principio , en cualquier compartimiento ,la probabilidad de que encuentre la comida antes de recibir una descarga es",p/7)





```

<h1><b><font color="red" >Solución del Problema 2 .Tarea 5 por Simulación de Monte Carlo</b></h1></font>


<p style="font-family: 'Arial', 'Helvetica', sans-serif; font-size: 16px; line-height: 1.6; color: #333333;">
  Se resolverá este problema utilizando simulación de Monte Carlo, entonces se realizará el código para hacer varias simulaciones, y se contará el número de tiradas que fueron necesarias  para teriminar el juego(en cada simulación) , luego se sumarán todas y se calculará el promedio, el cual será el número esperado de tiradas para terminar un juego solitario.
</p>

<p style="font-family: 'Arial', 'Helvetica', sans-serif; font-size: 16px; line-height: 1.6; color: #333333;">
  Algunas cuestiones a considerar son:
</p>

<ul style="font-family: 'Arial', 'Helvetica', sans-serif; font-size: 16px; line-height: 1.6; color: #333333; list-style-type: disc; margin-left: 20px;">
  <li>El juego comienza en la casilla $1$</li>
  <li>La probabilidad de que al lanzar el dado obtengamos un numero del $1$ al $6$, es $1/6$</li>
  <li>El juego termina al alcanzar la casilla $20$, como el avance es sin retorno , una vez que se llega o se supera la casilla final , se concluye.</li>
  <li>Hay $2$ escaleras, si la ficha cae en la casilla $3$, automaticamente sube a la casilla $11$. Si la ficha cae en la casilla $15$, entonces sube a la casilla $19$.</li>
  <li>Hay $2$ serpientes, si la ficha cae en la casilla $17$, baja a la $10$, Si la ficha cae en la casilla $13$, baja a la $4$.</li>
</ul>

<p style="font-family: 'Arial', 'Helvetica', sans-serif; font-size: 16px; line-height: 1.6; color: #333333;">
  Por tanto, los estados (Casillas) son
</p>

$$ S = \{ 1, 2, 4, 5, 6, 7, 8, 9, 10, 11, 12, 14, 16, 18, 19, 20 \} $$

<p style="font-family: 'Arial', 'Helvetica', sans-serif; font-size: 16px; line-height: 1.6; color: #333333;">
  No se incluyen el $3, 15, 17, 13$ porque al caer en ellas se realiza el movimiento de la escalera o serpiente.
</p>




Ahora procederemos a escribir el código ,en cada línea se explicará lo que se esta realizando.Para obtener una buena aproximación ,realizaremos N= 10000000  simulaciones.

```python
#Importamos la funcion 'choice' de la libreria 'random',la cual nos permitirá seleccionar aleatoriamente la casilla siguiente
from random import random,choice

#Definimos una función que nos regresará una lista de las casillas posibles a las que se puede mover la ficha , dependiendo de
#la casilla actual en la que se encuentra.
def casillas_posibles(casilla_actual):
  posibles=[]#lista vacía donde iremos guardando las casillas posibles
  resta=20-casilla_actual#calculamos el número de casillas que le faltan a la ficha para llegar al estado absorbente 20 y terminar el juego


#Ciclo para calcular las casillas posibles a las que puede avanzar la ficha ,dependiendo de la casilla  donde se encuentre,
#son 6 , ya que el dado tiene los números del 1 al 6, la 'i' representa los valores que el dado puede arrojar (1,2,3,4,5,6)
#para cada valor de 'i' , se agrega una casilla posible
  for i in range(1,7):
    #si el número que arroja el dado(i) es mayor o igual  que el número de casillas que faltan para llegar a la 20,entonces la ficha avanzará hasta la casilla 20
    if i>=resta:
      posibles.append(20)
    else:#si el valor del dado(i) no es suficiente para que la ficha avance hasta la número 20 ,entonces tenemos los siguiente:
      #'s' es la casilla posible a la que puede avanzar la ficha , dependiendo del valor del dado ('i')
      s=i+casilla_actual
      if s==3:
        #si s es la casilla 3 , la ficha no se quedará ahi(hay una escalera), avanzará a la casilla 11, entonces la casilla posible es 11 ,no 3
        posibles.append(11)
      elif s==15:
        #si s es la casilla 15 , la ficha no se quedará ahi(hay una escalera), avanzará a la casilla 19, entonces la casilla posible es 19 ,no 15
        posibles.append(19)
      elif s==17:
        #si s es la casilla 17 , la ficha no se quedará ahi(hay una serpiente), retrocederá a la casilla 10, entonces la casilla posible es 10 ,no 17
        posibles.append(10)
      elif s==13:
        #si s es la casilla 13 , la ficha no se quedará ahi(hay una serpiente), retrocederá a la casilla 4, entonces la casilla posible es 4 ,no 13
        posibles.append(4)
      else:
        #si s es una casilla  donde no hay una escalera o una serpiente , entonces la casilla posible es 's'
        posibles.append(s)
  return posibles#regresa una lista de las casillas posibles a las que puede avanzar la ficha , dependiendo de la casilla actual donde se encuentre

#Creamos una función que simulará UN JUEGO
def juego_escalera(casilla_inicial=1):#Hacemos que la casilla donde comience el juego sea la número 1
  casilla_actual=casilla_inicial#Definimos la casilla actual como la casilla inicial
  tiradas=0#contador que contará el número de tiradas hasta que termine el juego (se llegue a la casilla 20)
  while casilla_actual!=20:#Ciclo que simula las posiciones que va tomando la ficha durante este juego ,se detiene cuando la ficha toma la posicion 20
    tiradas+=1#se van sumando el número de tiradas ,hasta llegar a la casilla 20
    casilla_pasada=casilla_actual #hacemos que la casilla pasada sea la 'casilla actual' anteriormente definida
    #ahora ,nos fijamos en las casillas posibles a las que se puede mover la ficha(dependiendo de la casilla pasada) y seleccionamos una de forma aleatoria
    #cada una de las casillas posibles tiene la misma probabilidad
    casilla_actual=choice(casillas_posibles(casilla_pasada))
  return tiradas #devuelve el número de tiradas que fueron necesarias para terminar el juego (en UNA sola simulación) ,iniciando desde la casilla 1


N=10000000 #Es el número de simulaciones a realizar del juego , colocamos un número grande para tener una mejor precisión
suma=0  #Ira sumando el número de tiradas necesarias para terminar el juego (en cada simulación), iniciamos en 0 esta variable.
for i in range(N): #ciclo que realizará las N simulaciones
  #Se van  sumando el número de tiradas necesarias para terminar el juego (en cada simulación) si iniciamos en la casilla 1 .
  suma+=juego_escalera(casilla_inicial=1)

#Imprimimos el número de tiradas esperadas ,dividiendo suma/N
print('El numero de tiradas esperadas es',suma/N)
```

***Entonces obtenemos que el número de tiradas esperadas para terminar un juego solitario es : 6.8975315***
